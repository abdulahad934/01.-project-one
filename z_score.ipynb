```python
import numpy as np
from scipy.stats import norm
```

```python
sample = [172, 174, 168, 169, 171,
173, 175, 170, 169, 172]
population_mean = 170
population_std = 3
sample_mean = np.mean(sample)
n = len(sample)
```

```python
z_score = (sample_mean - population_mean) / (population_std / np.sqrt(n))
```

```python
z_score
```

```python
alpha = 0.05
```

```python
p_value = 2 * (1 - norm.cdf(abs(z_score)))
```

```python
p_value
```

```python
if p_value < alpha:
  print("i will reject the nul hypothesis")
else:
  print("i will accept the null hypothesis")
```
